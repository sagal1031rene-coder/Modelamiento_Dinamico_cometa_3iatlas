```python
#ejemplo de estructura para calcular la velocidad orbital
import numpy as np
def calcular_velocidad (distancia, masa_central):
  #G es la constante de gravitación universal
  G=6.67430e-11
  #v=sqrt(G*M/r) para órbitas circulares (ajustar para hiperbólicas)
  velocidad=np.sqrt(G*masa_central/distancia)
  return velocidad

#Datos iniciales
r=1.5e11 #Distancia en metros (ej. 1 UA)
m_sol=1.989e30 #Masa del Sol en kg

print(f"La velocidad calculada es: {calcular_velocidad(r, m_sol)} m/s")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
#Constantes en unidades astronómicas (UA) y días
G=4*np.pi**2 #Constante de gravitación en UA^3 / (Masa solsr* año*2) adaptada
#Simplificación: G en UA*3 /  (M_sol*día*2)
G_ua_dia=2.959122082855911e-04
M_sol=1.0

def ecuaciones_movimiento(t,estado, mu):
    x, y, vx, vy = estado
    r = np.sqrt(x**2 + y**2)
    #Aceleración gravitacional central (Sol)
    ax = -mu * x / r**3
    ay = -mu * y / r**3
    #NOTA: Aquí acoplaremos después el empuje por desgasificación de Marte
    return [vx, vy, ax, ay]
```

```python
#Cálculo de la Fuerza No Gravitatoria (empuje del Gas):
#Segunda ley de Newton en su forma de momento:
#F=dm/dt*v_gas
```

```python
#---CONFIGURACIÓN DE ESCENARIOS (Modifica estos valores según tus fuentes) ---
#Escenario 1: Publicación A
q1=1.4 #Distancia perihelio en UA (Cerca de la órbita de Marte)
# Assuming a value for q2 for the second scenario, and v2 for its velocity
q2 = 1.0 # Placeholder: Define q2
v1=1.6*np.sqrt(2*G_ua_dia*M_sol/q1) # Corrected to use q1 here, assuming v1 is for scenario 1
v2 = 1.6*np.sqrt(2*G_ua_dia*M_sol/q2) # Placeholder: Define v2 based on q2

#Condiciones iniciales [x, y, vx, vy] en el perihelo
estado1=[q1, 0, 0, v1]
estado2=[q2, 0, 0, v2]

t_span=(0, 100) #Simulación a 100 días post-perihelio
t_eval=np.linspace(t_span[0], t_span[1], 1000) # Corrected np.linespace to np.linspace and adjusted range

#Resolver
sol1=solve_ivp(ecuaciones_movimiento, t_span, estado1, args=(G_ua_dia*M_sol,), t_eval=t_eval) # Corrected '¨' to '*'
sol2=solve_ivp(ecuaciones_movimiento, t_span, estado2, args=(G_ua_dia*M_sol,), t_eval=t_eval)

#---GRAFICACIÓN---
plt.figure(figsize=(8,8))
plt.plot(sol1.y[0], sol1.y[1], label=f"Publicación A (q= {q1} UA)") # Corrected extra '('
plt.plot(sol2.y[0], sol2.y[1], label=f"Publicación B (q= {q2} UA)") # Corrected extra '('
# linestyle="_-" # This line seems misplaced if intended for plt.plot, and is not used
plt.scatter([0], [0], color='orange', s=200, label="Sol") # Corrected syntax for scatter and added quotes for color
plt.scatter([1.524], [0], color="red", s=50, label="Órbita prom. Marte (ref)") # Added quotes for label

plt.xlabel("X (UA)") # Corrected extra '"'
plt.ylabel("Y (UA)")
plt.title("Contraste de trayectoria hiperbólica para 3I/ATLAS") # Corrected typo in 'hiperbóloca'
plt.legend()
plt.grid(True)
plt.show()
```

este notebook presenta el análisis de ingeniería del objeto 3I/ATLAS aplicando balances de masa y energía paraexplicar su cambio de trayectoria. A diferencia de los modelos astrofísicos convencionales, se utiliza una densidad crítica de 500 kg/s para modelar el objeto como una estructura porosa sujeta a fallas mecánicas por presión de poros y fisión rotacional.

```python
#Visualización de Datos
import matplotlib.pyplot as plt
import numpy as np
#Curva de magnitud visualizando el salto de 18  17,2
#Vector de Empuje: Una gráfica de flechas mostrando cómo la liberación de gas a 30km/s cambia el vector de volocidad orbital.

```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

#---CONSTANTES Y UNIDADES---
G_ua_dia=2.959122082855911e-04
M_sol=1.0
#Conversiones útiles a Unidades Astronómicas (UA) y masas Solares (M:sol*dia^2)
Kg_A_MASA_SOLAR=1.0/1.989e30
METROS_A_UA=1.0/1.496e11

#---MIS DATOS (3I/ATLAS)---
diametro_km=2.6
radio_m=(diametro_km*1000)/2
densidad_estimada=500 # en kg/m^3 (ajustar si manejas otro valor)

#1. cálculo de la Masa Inicial (m0)
volumen_m3=(4/3)*np.pi*(radio_m**3)
masa_inicial_kg=volumen_m3*densidad_estimada
m0_sol=masa_inicial_kg*Kg_A_MASA_SOLAR

#2.Tasa de pérdida de masa (dm/dt)
#4.2 millones de toneladas por día=4.2e9 kg/dia
pérdida_masa_kg_dia=4.2e9
dm_dt_sol=pérdida_masa_kg_dia*Kg_A_MASA_SOLAR

print(f"Masa inicial estimada: {masa_inicial_kg:.4e} kg ({m0_sol:.4e} M_sol)")
print(f"Tasa de desgasificación: {pérdida_masa_kg_dia:.4e} kg/dia ({dm_dt_sol:.4e} M_sol/dia)")

#---ECUACIONES DE MOVIMIENTO CON MASA VARIABLE Y EMPUJE---
def ecuaciones_movimiento_variable(t, estado, mu, dm_dt_val):
    x, y, vx, vy, m_actual = estado

    #Si el objeto se queda sin masa (teórico), detenemos la pérdida
    if m_actual<=0:
      m_actual=1e-20 # Evitar división por cero
      dm_dt_val=0 # No hay más pérdida de masa

    r=np.sqrt(x**2+y**2)

    #Aceleración gravitacional del Sol

    ax_grav=-mu*x/r**3
    ay_grav=-mu*y/r**3

    #EFECTO COHETE (Fuerza No-Gravitacional por desgasificación)
    #Suponemos que el gas se expande principalmente en dirección opuesta al Sol
    # (radial)
    #V_escape típica de los gases cometarios aprox. 0.5 km/s -> convertida a UA/día
    v_escape_ua_dia=(500*86400)*METROS_A_UA # Convertir 500 m/s a UA/día

    #Fuerza=(dm/dt)*v_escape->aceleración=Fuerza/m_actual
    #El signo positivo empuja al objeto hacia a fuera del Sol
    # Corregir la fórmula de empuje, el factor 1/m_actual ya está en la aceleración
    ax_empuje = -(dm_dt_val / m_actual) * v_escape_ua_dia * x / r
    ay_empuje = -(dm_dt_val / m_actual) * v_escape_ua_dia * y / r

    #Derivadas: [dx/dt, dy/dt, dvx/dt, dvy/dt, dm/dt}
    return [vx, vy, ax_grav + ax_empuje, ay_grav + ay_empuje, -dm_dt_val]

#---SIMULACIÓN---
q=1.38 #Distancia al perihelio simulada (cercana a Marte)
v_inf=1.5*np.sqrt(2 * G_ua_dia * M_sol / q) #Velocdad hiperbólica

#Estado inicial: [x, y, vx, vy, masa)
estado_inicial=[q, 0, 0, v_inf, m0_sol]

t_span=(0, 100) #Simulación a 100 días de seguimiento (originalmente 150 días, ajustado para t_eval)
t_eval=np.linspace(t_span[0], t_span[1], 1500)

sol_con_desgasificacion=solve_ivp(ecuaciones_movimiento_variable,
                                  t_span,
                                  estado_inicial,
                                  args=(G_ua_dia*M_sol, dm_dt_sol),
                                  t_eval=t_eval, method='RK45')

#---GRAFICACIÓN---
plt.figure(figsize=(9, 9))
plt.plot(sol_con_desgasificacion.y[0], sol_con_desgasificacion.y[1],
         color="purple", label="Trayectoria 3i/ATLAS (Con desgasificación)")
plt.scatter([0], [0], color="orange", s=200, label="Sol")
plt.scatter([1.524], [0], color="red", s=70, label="Órbita de Marte (ref)")
plt.xlabel("X (UA)")
plt.ylabel("Y (UA)")
plt.title("Modelo de Ingeniería 3I/ATLAS: efecto de Pérdida de Masa")
plt.legend()
plt.grid(True)
plt.show()
```
